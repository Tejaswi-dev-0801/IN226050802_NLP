## Importing Required Modules

The following modules are used to load and operate the transformer model:

- `AutoTokenizer`: Responsible for converting text into tokenized input suitable for the model  
- `AutoModelForCausalLM`: Loads a pre-trained causal language model for response generation  
- `torch`: Supports tensor computations and model inference  

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

model_name = "gpt2"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)
```

## Chatbot Initialization

The chatbot is initialized with a welcome message to guide the user.  
A variable is defined to store conversation history, allowing the model to maintain contextual continuity across multiple interactions.

## Interactive Chat Loop and Response Generation

The chatbot operates in a continuous loop, enabling real-time interaction.

### Process Overview:
- Accept user input from the console  
- Evaluate exit conditions to terminate the session  
- Convert input text into tokens using the tokenizer  
- Append new input to the existing conversation history  
- Generate a response using the transformer model  
- Decode the generated tokens into readable text  
- Display the response to the user  

Maintaining conversation history ensures that responses remain contextually relevant.

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

model_name = "gpt2"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

# Fix padding issue
tokenizer.pad_token = tokenizer.eos_token

print("Chatbot: Hello! I am your AI assistant. How can I help you today?\n")

while True:
    user_input = input("User: ")

    if user_input.lower().strip() in ["exit", "quit"]:
        print("Chatbot: Chatbot ends the conversation")
        break

    prompt = f"""
You are a helpful AI assistant.

User: {user_input}
Chatbot:"""

    inputs = tokenizer(prompt, return_tensors="pt", padding=True)

    outputs = model.generate(
        inputs["input_ids"],
        attention_mask=inputs["attention_mask"],
        max_new_tokens=60,
        do_sample=True,
        temperature=0.5,     # less randomness
        top_k=40,
        top_p=0.9,
        repetition_penalty=1.2,
        pad_token_id=tokenizer.eos_token_id
    )

    response = tokenizer.decode(outputs[0], skip_special_tokens=True)

    # Extract only chatbot part
    response = response.split("Chatbot:")[-1].strip()

    print(f"Chatbot: {response}\n")
```
