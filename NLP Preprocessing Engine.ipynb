Task 1: Conceptual Understanding

1]Difference between "Love" and "love" in NLP
NLP treats words as tokens
Without normalization → "Love" and "love" are different
Lowercasing helps:
Reduce vocabulary size
Improve generalization
Reduce sparsity

2]What happens if stopwords are not removed?
Adds noise
Increases dimensionality
Slows training
Reduces performance (especially classification)

3]When removing stopwords is harmful

1. Sentiment Analysis
"I am not happy" → removing "not" changes meaning
2. Question Answering
"What is the capital of India?" → removing stopwords breaks structure

4]Stemming vs Lemmatization
Stemming: cuts words (fast, less accurate) using simple rules → studies → studi
Lemmatization: Converts words to their correct base form using a dictionary and context(accurate) → studies → study

```python
#Task 2: Advanced Preprocessing Function
import re

def preprocess_text(text):
    # Handle invalid input
    if not isinstance(text, str) or text.strip() == "":
        return [], ""

    # Lowercase
    text = text.lower()

    # Remove URLs & emails
    text = re.sub(r'http\S+|www\S+|https\S+', '', text)
    text = re.sub(r'\S+@\S+', '', text)

    # Remove numbers
    text = re.sub(r'\d+', '', text)

    # Handle repeated characters (soooo → soo)
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)

    # Remove emojis / special characters
    text = re.sub(r'[^a-zA-Z\s]', '', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    tokens = text.split()

    # Remove short tokens (<=2) except 'no', 'not'
    tokens = [w for w in tokens if len(w) > 2 or w in ["no", "not"]]

    clean_sentence = " ".join(tokens)

    return tokens, clean_sentence
```

```python
#Task3:Stress Testing
sample_inputs = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]

results = []

for text in sample_inputs:
    tokens, cleaned = preprocess_text(text)

    results.append({
        "original": text,
        "tokens": tokens,
        "cleaned": cleaned
    })

for r in results:
    print("Original:", r["original"])
    print("Tokens:", r["tokens"])
    print("Cleaned:", r["cleaned"])
    print("-" * 50)
```

```python
#Task 4: Token Analytics
import numpy as np

for r in results:
    tokens = r["tokens"]

    total = len(tokens)
    unique = len(set(tokens))
    avg_len = np.mean([len(t) for t in tokens]) if tokens else 0

    print("Sentence:", r["original"])
    print("Total Tokens:", total)
    print("Unique Tokens:", unique)
    print("Avg Token Length:", round(avg_len, 2))
    print("-" * 50)
```

```python
#Task 5: Frequency Analysis
from collections import Counter

all_tokens = []

for r in results:
    all_tokens.extend(r["tokens"])

counter = Counter(all_tokens)

print("Top 10 frequent words:")
print(counter.most_common(10))

print("\nTop 5 least frequent words:")
print(counter.most_common()[-5:])
```

```python
#Task 6: Full Pipeline
def full_pipeline(text_list):
    all_tokens = []
    clean_sentences = []

    for text in text_list:
        tokens, cleaned = preprocess_text(text)

        all_tokens.extend(tokens)
        clean_sentences.append(cleaned)

    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }

output = full_pipeline(sample_inputs)
print(output)
```
